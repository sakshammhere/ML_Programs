```python
df=pd.read_csv('kyphosis.csv')
```

```python
from sklearn.model_selection import train_test_split
```

```python
df.columns
```

```python
X=df[['Age', 'Number', 'Start']]
```

```python
y=df['Kyphosis']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3)
```

DecisionTreeClassifier

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dtree=DecisionTreeClassifier()
```

```python
dtree.fit(X_train,y_train)
```

```python
y_pred=dtree.predict(X_test)
```

```python
compare=pd.DataFrame({'Actual':y_test,'Predicted':y_pred})
```

```python
compare
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rndfrst=RandomForestClassifier()   #canuse n_estimators=200 or something here, try test which gives more accuracy
```

```python
rndfrst.fit(X_train,y_train)
```

```python
y_pred=rndfrst.predict(X_test)
```

```python
compare=pd.DataFrame({"Actual":y_test,"Predicted":y_pred})
```

```python
compare
```

```python
print(classification_report(y_test,y_pred))
```

In this situation , we can say that RandomForest works better and provides better accuracy as compared to a DecisionTree

In general, RandomForest gives better accuracy and precision as compared to DecisionTree
