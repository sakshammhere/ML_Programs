```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from catboost import CatBoostClassifier


# Save PassengerId
test_ids = test["PassengerId"]

# Combine for preprocessing
df = pd.concat([train, test], axis=0)

# =========================
# FEATURE ENGINEERING
# =========================

# Split Cabin
df[['Deck', 'CabinNum', 'Side']] = df['Cabin'].str.split('/', expand=True)
df.drop(columns=['Cabin'], inplace=True)

# Total spending
spend_cols = ['RoomService','FoodCourt','ShoppingMall','Spa','VRDeck']
df['TotalSpend'] = df[spend_cols].sum(axis=1)

# CryoSleep rule
df.loc[df['CryoSleep'] == True, spend_cols] = 0

# Group size
df['Group'] = df['PassengerId'].str.split('_').str[0]
group_size = df.groupby('Group')['PassengerId'].transform('count')
df['IsAlone'] = (group_size == 1).astype(int)

# Drop noise
df.drop(columns=['PassengerId','Name','Group'], inplace=True)

# =========================
# HANDLE MISSING VALUES
# =========================

df['Age'].fillna(df['Age'].median(), inplace=True)
df['VIP'].fillna(False, inplace=True)
df['CryoSleep'].fillna(False, inplace=True)

for col in ['HomePlanet','Destination','Deck','Side']:
    df[col].fillna(df[col].mode()[0], inplace=True)

for col in spend_cols + ['TotalSpend']:
    df[col].fillna(0, inplace=True)

# =========================
# ENCODE CATEGORICALS
# =========================

cat_cols = ['HomePlanet','CryoSleep','Destination','VIP','Deck','Side']

for col in cat_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])

# =========================
# SPLIT BACK
# =========================

X = df.iloc[:len(train)].drop(columns=['Transported'])
y = train['Transported'].astype(int)
X_test = df.iloc[len(train):].drop(columns=['Transported'])

# =========================
# MODEL
# =========================

model = CatBoostClassifier(
    iterations=800,
    depth=8,
    learning_rate=0.03,
    loss_function='Logloss',
    eval_metric='Accuracy',
    verbose=0
)

model.fit(X, y)

# =========================
# PREDICTION
# =========================

preds = model.predict(X_test)

submission = pd.DataFrame({
    'PassengerId': test_ids,
    'Transported': preds.astype(bool)
})

submission.to_csv("submission.csv", index=False)

print("submission.csv generated 🚀")
```
