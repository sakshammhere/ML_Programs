```python
train=pd.read_csv('titanic_train.csv')
```

```python
test=pd.read_csv('titanic_test.csv')
```

```python
train.head()
```

Cleaning the training data

```python
train.drop('PassengerId',axis=1,inplace=True)
```

```python
train.drop('Name',axis=1,inplace=True)
```

```python
train.info()
```

```python
train.drop('Cabin',axis=1,inplace=True)
```

```python
train['Age'].fillna(train['Age'].median(),inplace=True)
```

```python
train['Embarked'].fillna('Unknown',inplace=True)
```

```python
train=pd.get_dummies(train,columns=['Sex'],drop_first=True)
```

```python
train=pd.get_dummies(train,columns=['Embarked'],drop_first=True)
```

```python
bool_columns=train.select_dtypes(include='bool').columns
```

```python
train[bool_columns]=train[bool_columns].astype(int)
```

```python
train.head()
```

```python
train.drop('Embarked_Unknown',axis=1,inplace=True)
```

```python
train.head()
```

```python
#trained on the train data
X_train=train[['Pclass', 'Age', 'SibSp', 'Parch', 'Fare', 'Sex_male','Embarked_Q', 'Embarked_S']]
```

```python
y_train=train['Survived']
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.model_selection import train_test_split
```

```python
lm=LogisticRegression(max_iter=1000)
```

```python
lm.fit(X_train,y_train)
```

```python
#testing on the test data
y_pred=lm.predict(test)
```

```python
y_pred
```

```python
status = pd.Series(y_pred).map({1:'Survived', 0:'Not Survived'})
```

```python
test["Survival Status"]=status
```
