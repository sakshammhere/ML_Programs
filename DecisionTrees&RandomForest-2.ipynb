```python
cat_feats=['purpose']
```

```python
final_data=pd.get_dummies(loans,columns=cat_feats,drop_first=True)
```

```python
final_data.head()
```

```python
bool_cols=final_data.select_dtypes(include='bool').columns
```

```python
final_data[bool_cols]=final_data[bool_cols].astype(int)
```

```python
final_data.head()
```

```python
final_data.columns
```

```python
from sklearn.model_selection import train_test_split
```

```python
X=final_data[['credit.policy', 'int.rate', 'installment', 'log.annual.inc', 'dti','fico', 'days.with.cr.line',
               'revol.bal', 'revol.util','inq.last.6mths', 'delinq.2yrs', 'pub.rec',
       'purpose_credit_card', 'purpose_debt_consolidation','purpose_educational', 'purpose_home_improvement','purpose_major_purchase', 'purpose_small_business']]
```

```python
y=final_data['not.fully.paid']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=101)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dtree=DecisionTreeClassifier()
```

```python
dtree.fit(X_train,y_train)
```

```python
y_pred=dtree.predict(X_test)
```

```python
compare=pd.DataFrame({'Actual':y_test,"Predicted":y_pred})
```

```python
compare
```

```python
from sklearn.metrics import classification_report
```

```python
from sklearn.metrics import confusion_matrix
```

```python
print(classification_report(y_test,y_pred))
```

```python
print(confusion_matrix(y_test,y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rfc=RandomForestClassifier()   #n_estimators
```

```python
final_data.columns
```

```python
rfc.fit(X_train,y_train)
```

```python
y_pred1=rfc.predict(X_test)
```

```python
compare1=pd.DataFrame({"Actual":y_test,"Predictions":y_pred1})
```

```python
compare1
```

```python
print(classification_report(y_test,y_pred1))
```

```python
print(confusion_matrix(y_test,y_pred1))
```
